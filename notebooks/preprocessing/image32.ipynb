```python
import cv2
import csv
import os
labels={'Boot':0,'Sandal':1,'Shoe':2,'Sneaker':3}
l=list(labels.keys())

with open("Images32.csv",'w',newline='') as fl:
    for i in l:
        wr=csv.writer(fl)
        path = r'C:\My Files\ANN & ML Projects\ANNDL_Fall_2026\Project_ANNDL_Fall_2026\data\raw data\\'+i
        for r,d,f in os.walk(path):
            for fname in f:
                if '.jpg' or '.png' in fname:
                    record=[]
                    file_path=os.path.join(path,fname)
                    image1=cv2.imread(file_path)
                    image1=cv2.resize(image1,(32,32))
                    image1=cv2.cvtColor(image1,cv2.COLOR_RGB2GRAY)
                    image1=image1.flatten()
                    label1=labels[i]
                    record.append(label1)
                    record.extend(image1)
                    wr.writerow(record)

```
