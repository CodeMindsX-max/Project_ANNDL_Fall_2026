```python
# 1. MUST BE IMPORTED FIRST FOR KERAS 3 BACKEND
import tensorflow as tf
from keras.layers import Conv2D, Dense, Flatten, MaxPooling2D
from keras.models import Sequential
from keras.utils import to_categorical
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
# 2. Load custom CSV data
csv_path = r"/content/drive/MyDrive/100Images224.csv"
data = pd.read_csv(csv_path, header=None)

Y = data.iloc[:, 0].values
X = data.iloc[:, 1:].values

# Reshape flattened CSV features back to 224x224 RGB image tensors
X = X.reshape(-1, 224, 224, 3)
X = X.astype("float32") / 255.0

# 3. Train / Test Split
Xtrain, Xtest, Ytrain, Ytest = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

num_classes = 4
Ytrain = to_categorical(Ytrain, num_classes)
Ytest = to_categorical(Ytest, num_classes)

# 4. VGG-style CNN Model Architecture
cnn = Sequential()
cnn.add(
    Conv2D(
        filters=64,
        kernel_size=(2, 2),
        activation="relu",
        padding="same",
        input_shape=(224, 224, 3),
    )
)
cnn.add(MaxPooling2D(pool_size=(2, 2), strides=(2, 2)))

cnn.add(
    Conv2D(
        filters=128,
        kernel_size=(2, 2),
        activation="relu",
        padding="same",
        strides=(1, 1),
    )
)
cnn.add(MaxPooling2D((2, 2), strides=(2, 2)))

cnn.add(
    Conv2D(
        filters=256, kernel_size=(3, 3), activation="relu", padding="same"
    )
)
cnn.add(MaxPooling2D(pool_size=(2, 2), strides=(2, 2)))

cnn.add(
    Conv2D(
        filters=512,
        kernel_size=(3, 3),
        activation="relu",
        padding="same",
        strides=(1, 1),
    )
)
cnn.add(MaxPooling2D((2, 2), strides=(2, 2)))

cnn.add(
    Conv2D(
        filters=512,
        kernel_size=(3, 3),
        activation="relu",
        padding="same",
        strides=(1, 1),
    )
)
cnn.add(MaxPooling2D((2, 2), strides=(2, 2)))

cnn.add(Flatten())
cnn.add(Dense(4096, activation="relu"))
cnn.add(Dense(4096, activation="relu"))
cnn.add(Dense(4, activation="softmax"))

cnn.compile(
    loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"]
)

# 5. Train Model (Using batch_size=64 for stable training over 500 epochs)
result = cnn.fit(Xtrain, Ytrain, epochs=100, batch_size=32, verbose=1)

print(cnn.summary())

loss, acc = cnn.evaluate(Xtest, Ytest)
print("Loss: ", loss)
print("Accuracy: ", acc)

pd.DataFrame(result.history).plot()
plt.show()

# 6. Save Model to Drive Properly
cnn.save("/content/drive/MyDrive/100images224_vgg_model.keras")
print("Model successfully trained and saved to Google Drive!")
```
